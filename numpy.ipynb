```python
import numpy as np

arry = np.array([ 2, 4, 6 ])
print('array: ', arry)
```

```python
arry_2 = np.array([[22, 44, 66], [88, 00, 22]])
print('arry_2',arry_2.shape)
```

```python
a = np.array([12,2,33,4])
print('a_arryDtype: ', a.dtype)
```

```python
a = np.array([1, 3, 5, 7, 9, 11])
b = a.reshape(2, 3)

print('reshapedArray: ', b)
print('reshapedArray: ', b.shape)
```

```python
# astype()
a = np.array([1, 2, 3, 4])
b = a.astype(float)
print('float: ', b)
```

```python
# np.zeros() np.onese() => dummy data
zeros = np.zeros((1, 2, 3))
ones = np.ones((4, 5))
print('zeros: ', zeros)
print('ones: ', ones)
```

```python
# ndim
a = np.array([[1, 2, 3, 4], [1, 1, 2, 3]])
b = a.ndim
print('ndim: ', b)
```

```python
# arange()
a = np.arange(1, 9)
print('arange: ', a)
```

```python
# sum() mean() max() min()
a = np.array([1, 3, 5, 7, 9])
print('sum: ', a.sum())
print('mean: ', a.mean())
print('max: ', a.max())
print('min: ', a.min())
```

```python
#flatten

a = np.array([[ 1, 2, 3, 4 ], [ 5, 6, 7, 8 ]])
print(a.flatten())
```

```python
# T transpose

a = np.array([[1, 3, 4], [2, 6, 8]])
print("transpose: ", a.transpose())
print("transpose: ", a.T)
```

```python
#dot @

a = np.array([[2, 6], [1, 2]])
b = np.array([[3, 7], [3, 4]])

print('dot | multiplication: ', a.dot(b))
```

```python
#random rand

a = np.random.rand(3, 4)
print('random rand: ', a)
```
